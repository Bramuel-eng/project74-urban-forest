## Step 7: Column Consistency Check Across All Years
Verifying that all datasets have identical column structure for compatibility

```python
# Check column consistency

cols_2008 = set(dataset_2008.columns)
cols_2011 = set(dataset_2011.columns)
cols_2016 = set(dataset_2016.columns)
cols_2021 = set(dataset_2021.columns)

print("Column Consistency Check:")
print("="*70)

if cols_2008 == cols_2011 == cols_2016 == cols_2021:
    print("✓ All datasets have identical columns")
    print(f"\nTotal columns: {len(dataset_2008.columns)}")
else:
    print("⚠ Column differences detected\n")
    
    if cols_2008 != cols_2011:
        print("2008 vs 2011:")
        print(f"  Only in 2008: {cols_2008 - cols_2011}")
        print(f"  Only in 2011: {cols_2011 - cols_2008}\n")
    
    if cols_2008 != cols_2016:
        print("2008 vs 2016:")
        print(f"  Only in 2008: {cols_2008 - cols_2016}")
        print(f"  Only in 2016: {cols_2016 - cols_2008}\n")
    
    if cols_2008 != cols_2021:
        print("2008 vs 2021:")
        print(f"  Only in 2008: {cols_2008 - cols_2021}")
        print(f"  Only in 2021: {cols_2021 - cols_2008}")
```

## Step 8: Identify Area Column in Each Dataset
Find which column contains canopy area information

```python
# Check for area-related columns in each dataset

print("Area-related columns in each dataset:")
print("="*70)

# 2008
area_cols_2008 = [col for col in dataset_2008.columns 
                  if 'area' in col.lower() or 'shape' in col.lower()]
print(f"\n2008: {area_cols_2008}")

# 2011
area_cols_2011 = [col for col in dataset_2011.columns 
                  if 'area' in col.lower() or 'shape' in col.lower()]
print(f"2011: {area_cols_2011}")

# 2016
area_cols_2016 = [col for col in dataset_2016.columns 
                  if 'area' in col.lower() or 'shape' in col.lower()]
print(f"2016: {area_cols_2016}")

# 2021
area_cols_2021 = [col for col in dataset_2021.columns 
                  if 'area' in col.lower() or 'shape' in col.lower()]
print(f"2021: {area_cols_2021}")
```
